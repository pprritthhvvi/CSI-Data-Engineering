```python

# Apache Spark Introduction with Example

# What is Big Data?
# Big Data refers to extremely large datasets that cannot be
# processed efficiently using traditional databases.

# Why Apache Spark?
# Spark is a distributed data processing framework.
# It performs in-memory processing, making it much faster
# than Hadoop MapReduce.

# Features:
# 1. Fast
# 2. Distributed Processing
# 3. Fault Tolerant
# 4. Scalable
# 5. Supports SQL
# 6. Supports Machine Learning
# 7. Supports Streaming

# Import Spark

from pyspark.sql import SparkSession

# Create Spark Session

spark = SparkSession.builder \
    .appName("Spark Introduction") \
    .getOrCreate()

print("Spark Session Created Successfully")

# Create Sample Data

data = [
    (101,"Rahul","IT",70000),
    (102,"Priya","HR",50000),
    (103,"Amit","Finance",80000),
    (104,"Neha","IT",65000),
    (105,"Rohit","Finance",90000),
    (106,"Sneha","HR",55000),
    (107,"Karan","IT",75000),
    (108,"Pooja","HR",60000),
    (109,"Arjun","Finance",85000),
    (110,"Meena","IT",72000)
]

columns = ["ID","Name","Department","Salary"]

df = spark.createDataFrame(data, columns)

# Show Data

print("\nOriginal Data")
df.show()

# Print Schema

print("\nSchema")
df.printSchema()

# Describe Data

print("\nDescribe")
df.describe().show()

# Select Columns

print("\nSelect Name and Salary")
df.select("Name","Salary").show()

# Filter

print("\nSalary Greater Than 70000")
df.filter(df.Salary > 70000).show()

# Where

print("\nDepartment = IT")
df.where(df.Department == "IT").show()

# Order By

print("\nSort Salary Descending")
df.orderBy(df.Salary.desc()).show()

# GroupBy

print("\nAverage Salary Department Wise")

df.groupBy("Department").avg("Salary").show()

# Count

print("\nEmployee Count Department Wise")

df.groupBy("Department").count().show()

# Maximum Salary

print("\nMaximum Salary")

df.groupBy("Department").max("Salary").show()

# Minimum Salary

print("\nMinimum Salary")

df.groupBy("Department").min("Salary").show()

# Sum Salary

print("\nTotal Salary")

df.groupBy("Department").sum("Salary").show()

# Add New Column

from pyspark.sql.functions import col

print("\nBonus Column")

df.withColumn("Bonus", col("Salary")*0.10).show()

# Rename Column

print("\nRename Salary")

df.withColumnRenamed("Salary","MonthlySalary").show()

# Drop Column

print("\nDrop Department")

df.drop("Department").show()

# Remove Duplicate Rows

print("\nDistinct Departments")

df.select("Department").distinct().show()

# SQL Example

print("\nSpark SQL")

df.createOrReplaceTempView("employee")

spark.sql("""
SELECT Department,
AVG(Salary) AS AverageSalary
FROM employee
GROUP BY Department
""").show()

# Save Data

# Save as CSV
# df.write.csv("EmployeeOutput")

# Save as Parquet
# df.write.parquet("EmployeeParquet")

# Stop Spark Session

spark.stop()

print("Spark Session Stopped")
```
