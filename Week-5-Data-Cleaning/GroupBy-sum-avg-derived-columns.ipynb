Grouping Data with Pandas

. group by object

. basic Aggregate functions

· subgrouping

. agg() : Calculate multiple summary statistics for each group at once.

. apply() : Run a totally custom and complex function on each group.

. transform() : Perform a group calculation and return a result with the same shape as your original data.

· filter() : Keep or remove entire groups based on a condition.

```python
import pandas as pd

data = {
    "Name": ["Sneha","Pooja","Amit","Rohit","Meena",
             "Rahul","Ravi","Neha","Tina","Arjun",
             "Simran","Manish","Ankit","Priya","Karan"],

    "Department": ["IT","Marketing","HR","Marketing","Finance",
                   "IT","HR","HR","Marketing","Finance",
                   "IT","IT","Marketing","Finance","Finance"],

    "Salary": [78000,50000,45000,47000,68000,
               85000,48000,52000,53000,75000,
               80000,82000,55000,60000,72000]
}

df = pd.DataFrame(data)

print("Original Data")
print(df)

print("\nGroupBy")

for department, group in df.groupby("Department"):
    print("\nDepartment:", department)
    print(group)

print("\nSum")
print(df.groupby("Department")["Salary"].sum())

print("\nMean")
print(df.groupby("Department")["Salary"].mean())

print("\nMin")
print(df.groupby("Department")["Salary"].min())

print("\nMax")
print(df.groupby("Department")["Salary"].max())

print("\nCount")
print(df.groupby("Department")["Salary"].count())

print("\nAggregate Functions")
print(df.groupby("Department")["Salary"].agg(["sum","mean","min","max","count"]))

print("\nCustom Aggregate")
print(df.groupby("Department").agg(
    Total_Salary=("Salary","sum"),
    Average_Salary=("Salary","mean"),
    Minimum_Salary=("Salary","min"),
    Maximum_Salary=("Salary","max"),
    Employee_Count=("Salary","count")
))

print("\nSort by Total Salary")
print(df.groupby("Department")["Salary"].sum().sort_values(ascending=False))

print("\nFilter (Average Salary > 60000)")
print(df.groupby("Department").filter(lambda x: x["Salary"].mean() > 60000))

print("\nTransform")
df["Department_Average"] = df.groupby("Department")["Salary"].transform("mean")
print(df)

print("\nApply")
print(df.groupby("Department").apply(lambda x: x.nlargest(1, "Salary")))

print("\nMultiple GroupBy")
print(df.groupby(["Department","Name"])["Salary"].sum())

print("\nDescribe")
print(df.groupby("Department")["Salary"].describe())
```

```python
import pandas as pd

data = {
    "Employee": ["Rahul","Priya","Amit","Neha","Rohit","Sneha","Karan","Pooja","Arjun","Meena"],
    "Department": ["IT","IT","HR","HR","Finance","Finance","IT","HR","Finance","IT"],
    "Salary": [70000,65000,50000,55000,80000,75000,72000,52000,82000,68000],
    "Experience": [3,5,2,4,6,5,4,3,7,2]
}

df = pd.DataFrame(data)

print(df)

print("\nSubgrouping")
print(df.groupby(["Department", "Experience"]).sum(numeric_only=True))

print("\nagg()")
print(df.groupby("Department").agg(
    Total_Salary=("Salary","sum"),
    Average_Salary=("Salary","mean"),
    Maximum_Salary=("Salary","max"),
    Minimum_Salary=("Salary","min"),
    Employee_Count=("Salary","count")
))

print("\napply()")
print(df.groupby("Department").apply(lambda x: x.nlargest(1, "Salary")))

print("\ntransform()")
df["Department_Average"] = df.groupby("Department")["Salary"].transform("mean")
print(df)

print("\nfilter()")
print(df.groupby("Department").filter(lambda x: x["Salary"].mean() > 70000))
```

```python
import pandas as pd

data = {
    "Employee": ["Rahul","Priya","Amit","Neha","Rohit","Sneha","Karan","Pooja","Arjun","Meena"],
    "Department": ["IT","IT","HR","HR","Finance","Finance","IT","HR","Finance","IT"],
    "Salary": [70000,65000,50000,55000,80000,75000,72000,52000,82000,68000],
    "Experience": [3,5,2,4,6,5,4,3,7,2]
}

df = pd.DataFrame(data)

print("Original Data")
print(df)

print("\nGroupBy")
for dept, group in df.groupby("Department"):
    print("\nDepartment:", dept)
    print(group)

print("\nAggregation")
print(df.groupby("Department")["Salary"].agg(["sum","mean","min","max","count"]))

print("\nTransformation")
df["Average_Salary"] = df.groupby("Department")["Salary"].transform("mean")
print(df)

print("\nFiltration")
print(df.groupby("Department").filter(lambda x: x["Salary"].mean() > 70000))

print("\nApply")
print(df.groupby("Department").apply(lambda x: x.nlargest(1, "Salary")))

print("\nSubgroups")
print(df.groupby(["Department","Experience"])["Salary"].sum())
```
