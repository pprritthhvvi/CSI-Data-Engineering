```python
import pandas as pd
import numpy as np

data = {
    "EmployeeID": [101,102,103,104,105,106,107,108],
    "Name": ["Rahul","Priya","Amit",None,"Sneha","Rahul","Karan","Pooja"],
    "Department": ["IT","HR","IT","Finance","HR",None,"IT","Finance"],
    "Salary": [70000,50000,np.nan,80000,55000,70000,72000,85000],
    "Age": [25,28,26,np.nan,30,25,27,29],
    "City": ["Delhi","Mumbai","Delhi","Pune","Mumbai","Delhi","Chennai","Pune"]
}

df = pd.DataFrame(data)

print("Original Data")
print(df)

print("\nMissing Values")
print(df.isnull().sum())

df["Name"] = df["Name"].fillna("Unknown")
df["Department"] = df["Department"].fillna("Not Assigned")
df["Salary"] = df["Salary"].fillna(df["Salary"].mean())
df["Age"] = df["Age"].fillna(df["Age"].mean())

print("\nAfter Filling Missing Values")
print(df)

df = df.drop_duplicates()

print("\nAfter Removing Duplicates")
print(df)

df = df.rename(columns={"Salary":"MonthlySalary"})

print("\nAfter Renaming Column")
print(df)

df["Age"] = df["Age"].astype(int)

print("\nAfter Changing Data Type")
print(df.dtypes)

df["Bonus"] = df["MonthlySalary"] * 0.10

print("\nBonus Column")
print(df)

df["SalaryCategory"] = np.where(
    df["MonthlySalary"] >= 70000,
    "High",
    "Low"
)

print("\nSalary Category")
print(df)

print("\nSort by Salary")
print(df.sort_values("MonthlySalary", ascending=False))

print("\nEmployees with Salary > 70000")
print(df[df["MonthlySalary"] > 70000])

print("\nSelected Columns")
print(df[["Name","Department","MonthlySalary"]])

print("\nAverage Salary Department Wise")
print(df.groupby("Department")["MonthlySalary"].mean())

print("\nAggregate")
print(
    df.groupby("Department")["MonthlySalary"].agg(
        ["sum","mean","min","max","count"]
    )
)

df["DepartmentAverage"] = (
    df.groupby("Department")["MonthlySalary"]
      .transform("mean")
)

print("\nTransform")
print(df)

print("\nApply")
print(
    df.groupby("Department")
      .apply(lambda x: x.nlargest(1,"MonthlySalary"))
)

print("\nFilter")
print(
    df.groupby("Department")
      .filter(lambda x: x["MonthlySalary"].mean() > 65000)
)

print("\nSummary Statistics")
print(df.describe())

print("\nFinal Cleaned Data")
print(df)
```
